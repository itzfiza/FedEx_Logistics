# **Project Summary -**

This project is an Exploratory Data Analysis (EDA) on the FedEx Logistics shipment dataset. The dataset contains **10,324 shipment records** and **33 columns** representing purchase order (PO) details, shipment modes, vendor INCO terms, scheduled and actual delivery dates, product specifications, weight, freight, and insurance costs.

Our key focus was to understand the driving factors behind on-time delivery performance (KPI) and shipment cost structures to help the logistics team optimize delivery speed and reduce unnecessary costs.

### **Key Steps Taken:**
1. **Data Cleaning & Wrangling:** Converted five date columns and all numerical fields to correct datatypes. Handled implicit missing values hidden under descriptive texts (e.g., *'Pre-PQ Process'*, *'Date Not Captured'*, *'Freight Included'*) by coercing them to NaT/NaN. 4. **Statistical Testing:** Conducted Chi-Square, Mann-Whitney U, and Spearman Rank Correlation tests to prove or disprove operational assumptions.

### **Main Findings & Impact:**
* **Delivery Baseline:** **88.5%** of shipments are delivered **On Time**, while **11.5%** are **Delayed**.
* **Operational Bottlenecks:** Shipment mode and destination country significantly affect delays. Air has the highest on-time rate (**90.4%**), whereas Ocean has the lowest (**82.5%**). Among top destinations, **Mozambique** has the highest delay rate (**18.4%**).
* **Lead Time Myth:** A Mann-Whitney U test proved that transit lead time has **no statistically significant relationship** with delivery delays ($p = 0.4122$).
* **Cost Drivers:** Air transportation is the most expensive carrier option (median **$10.02/kg**), while Ocean is highly cost-effective (median **$1.68/kg**). Insurance costs are almost perfectly correlated with cargo value (Spearman $\rho = 0.99$) rather than physical weight.

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df_clean = df.copy()                                   # keep the original safe
df_clean.columns = df_clean.columns.str.strip()        # remove extra spaces in column names

DATE_COLS = ["PQ First Sent to Client Date", "PO Sent to Vendor Date", "Scheduled Delivery Date",
             "Delivered to Client Date", "Delivery Recorded Date"]
NUM_COLS = ["Unit of Measure (Per Pack)", "Line Item Quantity", "Line Item Value", "Pack Price",
            "Unit Price", "Weight (Kilograms)", "Freight Cost (USD)", "Line Item Insurance (USD)"]


def convert_dates(series):
    """Convert text like 02-Jun-06 to datetime. Text like 'Pre-PQ Process' becomes NaT."""
    parsed = pd.to_datetime(series, format="%d-%b-%y", errors="coerce")
    left = parsed.isna() & series.notna()              # try a general parse for odd formats
    if left.any():
        parsed[left] = pd.to_datetime(series[left], errors="coerce", dayfirst=True)
    return parsed


def convert_numeric(series):
    """Convert to number. Text like 'Invoiced Separately' becomes NaN."""
    return pd.to_numeric(series, errors="coerce")


# 1. text placeholders hiding inside the numeric columns (before converting)
for col in ["Weight (Kilograms)", "Freight Cost (USD)"]:
    if col in df_clean.columns:
        bad = df_clean[col][convert_numeric(df_clean[col]).isna() & df_clean[col].notna()]
        print(f"{col}: {len(bad)} text values, e.g. {bad.unique()[:2].tolist()}")

# 2. convert date and numeric columns
for col in DATE_COLS:
    if col in df_clean.columns:
        df_clean[col] = convert_dates(df_clean[col])
for col in NUM_COLS:
    if col in df_clean.columns:
        df_clean[col] = convert_numeric(df_clean[col])

# 3. missing categorical values -> label (numeric ones are kept as NaN on purpose)
df_clean["Shipment Mode"] = df_clean["Shipment Mode"].fillna("Unknown")
df_clean["Dosage"] = df_clean["Dosage"].fillna("Not Specified")
for col in df_clean.select_dtypes(include="object").columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

# 4. remove duplicates if any
before = df_clean.shape[0]
df_clean = df_clean.drop_duplicates().reset_index(drop=True)
print("Duplicates removed:", before - df_clean.shape[0])

# 5. new columns (feature engineering)
df_clean["Delay Days"] = (df_clean["Delivered to Client Date"] - df_clean["Scheduled Delivery Date"]).dt.days
df_clean["Delivery Status"] = np.select(
    [df_clean["Delay Days"].isna(), df_clean["Delay Days"] <= 0],
    ["Unknown", "On Time"], default="Delayed")         # delivered on/before schedule = On Time
df_clean["Lead Time (Days)"] = (df_clean["Scheduled Delivery Date"] - df_clean["PO Sent to Vendor Date"]).dt.days
df_clean["Freight per Kg"] = df_clean["Freight Cost (USD)"] / df_clean["Weight (Kilograms)"].replace(0, np.nan)
df_clean["Delivery Month"] = df_clean["Delivered to Client Date"].dt.to_period("M").dt.to_timestamp()

# log columns for skewed data (used in charts)
for col in ["Line Item Quantity", "Line Item Value", "Weight (Kilograms)", "Freight Cost (USD)"]:
    df_clean[col + " (log)"] = np.log1p(df_clean[col].clip(lower=0))

# 6. outlier check using IQR (not removing them, they are real bulk orders)
def count_outliers_iqr(s):
    s = s.dropna()
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    return int(((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum())

outlier_df = pd.DataFrame({
    "Outliers (IQR)": {c: count_outliers_iqr(df_clean[c]) for c in NUM_COLS},
    "Valid values": {c: int(df_clean[c].notna().sum()) for c in NUM_COLS}})
display(outlier_df)

# 7. data only with a known delivery status (used for delay analysis)
df_status = df_clean[df_clean["Delivery Status"] != "Unknown"].copy()
top_countries = df_clean["Country"].value_counts().head(10).index.tolist()

print("\nShape after cleaning:", df_clean.shape)
print("Rows with known delivery status:", df_status.shape[0])
display(df_clean["Delivery Status"].value_counts())
display(df_clean[NUM_COLS + DATE_COLS].isnull().sum().to_frame("Missing after cleaning"))
````

````python
# Helper functions used in the charts (written once, used many times)
STATUS_COLORS = {"On Time": "#2a9d8f", "Delayed": "#e76f51"}
MAIN_COLOR = "#3b6ea5"
TEST_RESULTS = {}      # all test results are stored here and shown in section 5


def chi_square_test(data, col, name):
    """Chi-square test between a categorical column and delivery status."""
    ct = pd.crosstab(data[col], data["Delivery Status"])
    if ct.shape[0] < 2 or ct.shape[1] < 2:
        print("Not enough categories for chi-square test")
        return None
    chi2, p, _, _ = stats.chi2_contingency(ct)
    TEST_RESULTS[name] = {"Test": "Chi-square", "Statistic": round(chi2, 2), "p-value": round(p, 4),
                          "Significant (p<0.05)": "Yes" if p < 0.05 else "No"}
    print(f"Chi-square test ({name} vs delivery status): chi2 = {chi2:.2f}, p-value = {p:.4f}")
    return p


def on_time_rate_plot(data, col, title, top_n=None, rotate=0):
    """Bar chart of on-time delivery % for every category of a column + chi-square test."""
    d = data.copy()
    if top_n:
        d = d[d[col].isin(d[col].value_counts().head(top_n).index)]
    summary = d.groupby(col).agg(
        Shipments=("Delivery Status", "size"),
        OnTimePct=("Delivery Status", lambda s: (s == "On Time").mean() * 100)
    ).sort_values("OnTimePct", ascending=False)

    plt.figure(figsize=(10, 5))
    ax = plt.gca()
    ax.bar(summary.index.astype(str), summary["OnTimePct"], color=MAIN_COLOR)
    ax.bar_label(ax.containers[0], fmt="%.1f%%", padding=2, fontsize=9)
    ax.axhline((d["Delivery Status"] == "On Time").mean() * 100, color="red", ls="--", label="Overall on-time %")
    ax.set_ylim(0, 115)
    ax.set_ylabel("On-time deliveries (%)")
    ax.set_title(title)
    ax.legend()
    plt.xticks(rotation=rotate, ha="right" if rotate else "center")
    plt.tight_layout()
    plt.show()
    chi_square_test(d, col, col)
    display(summary.round(1))


def spearman_test(x, y, name):
    """Spearman correlation (good for skewed data), result stored in TEST_RESULTS."""
    rho, p = stats.spearmanr(x, y)
    TEST_RESULTS[name] = {"Test": "Spearman", "Statistic": round(rho, 3), "p-value": round(p, 4),
                          "Significant (p<0.05)": "Yes" if p < 0.05 else "No"}
    print(f"Spearman correlation ({name}): rho = {rho:.3f}, p-value = {p:.4f}")
````

#### Chart - 13

````python
# Chart - 13 visualization code
lt = df_status.dropna(subset=["Lead Time (Days)"])
lt = lt[lt["Lead Time (Days)"] >= 0]
order = ["On Time", "Delayed"]
groups = [lt.loc[lt["Delivery Status"] == s, "Lead Time (Days)"] for s in order]

plt.figure(figsize=(7, 5))
bp = plt.boxplot(groups, patch_artist=True, showfliers=False)
for patch, s in zip(bp["boxes"], order):
    patch.set_facecolor(STATUS_COLORS[s])
plt.xticks([1, 2], order)
plt.ylabel("Lead time: PO sent to scheduled delivery (days)")
plt.title("Lead time vs delivery status")
plt.tight_layout()
plt.show()

# Mann-Whitney test (lead time is skewed so no t-test)
if all(len(g) > 0 for g in groups):
    u, p = stats.mannwhitneyu(groups[0], groups[1])
    TEST_RESULTS["Lead time (On Time vs Delayed)"] = {"Test": "Mann-Whitney U", "Statistic": round(u, 1),
                                                       "p-value": round(p, 4),
                                                       "Significant (p<0.05)": "Yes" if p < 0.05 else "No"}
    print(f"Median lead time -> On Time: {groups[0].median()} days, Delayed: {groups[1].median()} days")
    print(f"Mann-Whitney U test: p-value = {p:.4f}")
````

##### 1. Why did you pick the specific chart?

A box plot compares the spread and median of a numeric column between two groups. Outliers are hidden to keep the boxes readable.

##### 2. What is/are the insight(s) found from the chart?

On-time shipments have a median lead time of **111.0 days**, while delayed shipments have a median lead time of **115.0 days**. The Mann-Whitney U test shows **no statistically significant difference** in lead times between the two groups ($p = 0.4122 \ge 0.05$). This indicates that lead time is not a major factor causing delays.

##### 2. What is/are the insight(s) found from the chart?
The median lead time of on-time shipments is 111.0 days, while for delayed shipments it is 115.0 days. Since the Mann-Whitney U test p-value is 0.4122, this difference is not statistically significant (p > 0.05), suggesting that lead time alone does not determine delay status.

##### 3. Will the gained insights help creating a positive business impact?

* **Positive impact:** Since lead time has no significant difference, the logistics team should focus on route operations, customs clearing, and carrier selection rather than simply extending target timelines.
